```python
import gensim
import os

```

```python
from nltk import sent_tokenize
from nltk.corpus import stopwords

STOPWORDS = set(stopwords.words("english"))
from gensim.utils import simple_preprocess

story_line = []

with open("dataSet/Friends.csv", "r", encoding="utf-8") as f:
    corpus = f.read()

sentences = sent_tokenize(corpus)
story = [
    [t for t in simple_preprocess(s) if t not in STOPWORDS]
    for s in sentences
]
```

```python
import gensim
from gensim.models import Word2Vec
model = Word2Vec(
    sentences=story,
    vector_size=100,
    window=5,
    min_count=1,
    workers=os.cpu_count() or 1,
    sg=1,
)

# training already happened in the constructor above
print("vocab size:", len(model.wv))
print(model.wv.most_similar("friend", topn=5))
```

```python
model.train(story, total_examples=model.corpus_count, epochs=model.epochs)
```
